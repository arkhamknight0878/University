```python
# Подключаем библиотеки
import sympy as sym
import numpy as np
import sympy.plotting.plot as symplot
from IPython.display import display, Math
from sympy import init_printing
import matplotlib.pyplot as plt
```

##ЗАДАНИЕ 1

```python
a = sym.symbols('a')
```

```python
matrix_sym = sym.Matrix([[1, -a, 0], [-a, -2, 1], [0, 1, -2]])
display(matrix_sym)
```

Посчитаем детерминант

```python
matrix_det = matrix_sym.det()
print()
display(matrix_det)
```

Транспонируем матрицу

```python
matrix_transpon = matrix_sym.transpose()
display(matrix_transpon)
```

Теперь найдём обратную матрицйу

```python
print("Обратная матрица")

matrix_inv = matrix_sym.inv()

print()
display(matrix_inv)
```

Умножим обратную на обычную и упростим выражения

```python
print ("Результат умножения обычной матрицы на обратную без упрощения выражений")
inv_check = matrix_inv * matrix_sym

display(inv_check)

print()
print("Результат умножения обычной матрицы на обратную с упрощением выражений")

display(inv_check.applyfunc(sym.simplify))
```

Получилась единичная матрица ⇒ обратная матрица была найдена верно

##ЗАДАНИЕ 2

Задаём матрицы коэффициентов и расширенные матрицы для систем 2), 3), 4)

```python
slau1 = sym.Matrix([[1, -1, 0], [-1, -2, 1], [0, 1, -2]])
slau2 = sym.Matrix([[1, -1, 0], [-1, -2, 1], [0, -3, 1]])
slau3 = sym.Matrix([[1, -1, 0], [-1, -2, 1], [0, -3, 1]])

slau1_ex = sym.Matrix([[1, -1, 0, 1], [-1, -2, 1, 2], [0, 1, -2, 0]])
slau2_ex = sym.Matrix([[1, -1, 0, 1], [-1, -2, 1, 2], [0, -3, 1, 2]])
slau3_ex = sym.Matrix([[1, -1, 0, 1], [-1, -2, 1, 2], [0, -3, 1, 3]])
```

```python
print("Матрица коэффициентов и её расширенная версия для системы 1:\nНазовём их (1) и (1.1) соответственно")
display(slau1)
print()
display(slau1_ex)

print("\nМатрица коэффициентов и её расширенная версия для системы 2:\nНазовём их (2) и (2.1) соответственно")
display(slau2)
print()
display(slau2_ex)

print("\nМатрица коэффициентов и её расширенная версия для системы 3:\nНазовём их (3) и (3.1) соответственно")
display(slau3)
print()
display(slau3_ex)
```

Находим ранги для заданных матриц

```python
sl1_rnk = slau1.rank()
sl2_rnk = slau2.rank()
sl3_rnk = slau3.rank()

sl1ex_rnk = slau1_ex.rank()
sl2ex_rnk = slau2_ex.rank()
sl3ex_rnk = slau3_ex.rank()
```

```python
print("Ранги матриц (1) и (1.1) соответственно:", sl1_rnk, ",", sl1ex_rnk)
print("Ранги матриц (2) и (2.1) соответственно:", sl2_rnk, ",", sl2ex_rnk)
print("Ранги матриц (3) и (3.1) соответственно:", sl3_rnk, ",", sl3ex_rnk)
```

Ранги матриц (2) и (2.1) не совпадают, поэтому, по теореме Кронекера-Капелли, система 2 не имеет решений. Решим системы 1 и 3:

```python
x1, x2, x3 = sym.symbols('x_1, x_2, x_3')

roots1 = sym.linsolve(slau1_ex, (x1, x2, x3))
roots3 = sym.linsolve(slau3_ex, (x1, x2, x3))
```

```python
print("Решение системы уравнений 1:")
display(roots1)
print("\nРешение системы уравнений 3:")
display(roots3)
```

```python
eigenvals_subs = new_matrix.eigenvals()
eigenvects_subs = new_matrix.eigenvects()

print("Собственные значения:\n")
display(Math(sym.latex(eigenvals_subs)))

print("\nСобственные векторы:\n")
display(Math(sym.latex(eigenvects_subs)))
```

Найдём собственные значения, используя команды sympy:

Построим матрицу A - λE:

```python
lambda_sym = sym.symbols('lambda')
eimatrix = new_matrix - lambda_sym * inv_check.applyfunc(sym.simplify)
display(eimatrix)
```

Построим характерестический многочлен это матрицы:

```python
char_pol = eimatrix.det()
display(char_pol)
```

Решим получившееся уравнение:

```python
eigenvals_sym = sym.solve(char_pol)

print("Получившиеся собственные числа:\n")
display(Math(sym.latex(eigenvals_sym)))
```

Для первого корня найдём собственный вектор:

##ЗАДАНИЕ 1

Даны функции:

```python
K, x = sym.symbols('K, x')
f_x = K * sym.cos(3 * x) * sym.exp(-4 * x)
```

Ищем производные функций:

```python
df_dx = sym.diff(f_x, x)
df2_dx2 = sym.diff(f_x, x, 2)

print("Первая производная:\n")
display(df_dx)
print("\nВторая производная:\n")
display(df2_dx2)
```

Ищем интегралы:

```python
indefinite_integral = sym.integrate(f_x, x)
definite_integral = sym.integrate(f_x, (x, 0, sym.pi))

print("\nНеопределённый интеграл:\n")
display(indefinite_integral)

print("\nОпределённый интеграл от 0 до pi:\n")
display(definite_integral)
```

Ищем пределы:

```python
limit_at_2 = sym.limit(f_x,x,2)
limit_at_zero = sym.limit(f_x, x, 0)
limit_at_infinity = sym.limit(f_x, x, sym.oo)

print("Предел при x -> 2:\n")
display(limit_at_2)

print("\nПредел при x -> 0:\n")
display(limit_at_zero)

print("\nПредел при x -> ∞:\n")
display(limit_at_infinity)
```

##ЗАДАНИЕ 2

Ряд Тейлора с членами до 2-й степени

```python
x = sym.symbols('x')
```

```python
K_value = 1
x0 = 1

taylor_linear = f_x.series(x, x0, 2).removeO()
taylor_quadratic = f_x.series(x, x0, 2).removeO()

print("Линейная аппроксимация:")
display(taylor_linear)
print("Квадратичная аппроксимация:")
display(taylor_quadratic)
```

Построим график

```python
K_value = 1
f_num = sym.lambdify(x, f_x.subs({K: K_value}), modules=['numpy'])
taylor_linear_num = sym.lambdify(x, taylor_linear.subs({K: K_value}), modules=['numpy'])
taylor_quadratic_num = sym.lambdify(x, taylor_quadratic.subs({K: K_value}), modules=['numpy'])
```

Настроим цвет, толщину и т.п. и начертим график

```python
x0 = 0.5
x_vals = np.linspace(x0 - 0.5, x0 + 0.5, 200)
plt.figure(figsize=(10, 6))
plt.plot(x_vals, f_num(x_vals), label='f(x) = K*cos(3x)*exp(-4x)', color='blue', linewidth=2)
plt.plot(x_vals, taylor_linear_num(x_vals), label='Линейная аппроксимация', color='red', linestyle='--', linewidth=1.5)
plt.plot(x_vals, taylor_quadratic_num(x_vals), label='Квадратичная аппроксимация', color='green', linestyle=':', linewidth=1.5)

plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Функция Kcos(3x)*exp(-4x) и её аппроксимации ряда Тейлора в точке x = 0.5')
plt.legend()
plt.grid(True)
plt.ylim([-1, 1])
plt.xlim([x0 - 0.5, x0 + 0.5])

plt.show()
```

##ЗАДАНИЕ 3

```python
x = sym.symbols('x')
gx = (x**2 + 5*x + 7) / (x + 2)
```

Найдём вертикальные асимптоты:

Область определения: x не равен -2. Т.е. точка x = -2 - точка разрыва.
Убедимся в том, что x = -2 - вертикальная асимптота:

```python
lim_pnt = -2
lim = sym.limit(gx, x, lim_pnt)
display(lim)
```

Предел равен бесконечности, это означает, что
функция в этой точке бесконечно приближается к линии асимптоты.
Т.е. x = -2 - вертикальная асимптота.

Теперь найдём горизонтальные асимптоты:

Сначала рассчитаем коэффициенты k и b:

```python
k = sym.limit(gx / x, x, sym.oo)
b = sym.limit(gx - k*x, x, sym.oo)
```

```python
print("k = ", k)
print("b = ", b)
```

k не равно 0 или oo, значит y = x + 3 - наклонная асимптота c углом наклона 1.

```python
vert_asym = x + 3
```

Теперь, используя производную, найдём критические точки:

```python
dg = sym.diff(gx, x)
crit_points = sym.solve(dg, x)
```

```python
print("Получившаяся производная:")
display(Math(sym.latex(dg)))

print("\nНайденые критические точки:")
display(crit_points)
```

Используя вторую производную, определим тип экстремума в критических точках:

```python
dg2 = sym.diff(dg, x)

extrema_points = []
extrema_values = []
extrema_types = []

for point in crit_points:
    extrema_points.append(point)
    extrema_values.append(gx.subs({x: point}))
    second_derivative_value = dg2.subs({x: point})
    if second_derivative_value > 0:
        extrema_types.append("Минимум")
    elif second_derivative_value < 0:
        extrema_types.append("Максимум")
    else:
        extrema_types.append("Точка перегиба")
```

```python
print("Получившаяся вторая производная:")
display(Math(sym.latex(dg2)))

print("\nНайденные точки экстремума (x): ", extrema_points)
print("Значения в точках экстремума (y): ", extrema_values)
print("Типы точек экстремума: ", extrema_types)
```

Теперь построим график:

```python
grfph = symplot(gx, (x, -7, 7), show = False)
grfph.extend(symplot(vert_asym, (x, -7, 7), show = False))

grfph[0].label = '$g(x) = %s$' %sym.latex(gx)
grfph[1].label = '$d(x) = %s$' %sym.latex(vert_asym)

grfph.ylim = [-7, 7]
grfph.xlim = [-7, 4]

grfph[0].line_color = 'r'
grfph[1].line_color = 'b'

grfph.legend = True

grfph.show()
```
